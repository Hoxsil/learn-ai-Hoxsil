# step 1:导入库

```python
import matplotlib.pyplot as plt
import numpy as np
```

# step 2: 

```python
x = np.array([2022, 2023, 2024, 2025])
y1 = np.array([1, 2, 3, 4])
y2 = np.array([2, 3, 1, 2])
# 创建标题
plt.title("class, sise", fontsize=25,
                         family="Arial",
                         fontweight="bold",
                         color='#12f1a2')
# 创建 x 轴标签
plt.xlabel("Year", fontsize=20,
                   family="Arial",
                   fontweight="bold",
                   color="red")
# 定义 x 轴刻度
plt.xticks(x)
# 加入一组图表数据
plt.plot(x, y1, marker='.',
                markersize=7,
                markerfacecolor="yellow",
                markeredgecolor="red",
                linestyle="dashed",
                linewidth=4,
                color="cyan")
# linestyle=dashed solid dashdot dotted dashed None
plt.plot(x, y2, linestyle="dotted")
# 显示图标
plt.show()
```
